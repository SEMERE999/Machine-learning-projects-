```python
import pandas as pd
import numpy as np 
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import load_breast_cancer
```

```python
bc=load_breast_cancer()
x=pd.DataFrame(bc.data , columns=bc.feature_names)
#x['target']=bc.target
y=pd.Series(bc.target)
```

```python
x.duplicated().sum()
x.T.duplicated().sum()
```

```python
x.columns=x.columns.str.replace(' ','_')
x.columns
```

```python
x.describe()
```

```python
from sklearn.preprocessing import StandardScaler
z=StandardScaler()
x_trns=z.fit_transform(x)
x=pd.DataFrame(x_trns,columns=x.columns)
x.head(10)
```

```python
x.describe()
```

```python

from sklearn.model_selection import train_test_split
x1,xt,y1,yt=train_test_split(x,y,test_size=0.2,random_state=42)
log=LogisticRegression(max_iter=1000)
log.fit(x,y)
pred=log.predict(xt)

```

```python
y_proba = log.predict_proba(xt)[:, 1]
from sklearn.metrics import accuracy_score, confusion_matrix

print("Accuracy:", accuracy_score(yt, pred))
print("Confusion Matrix:\n", confusion_matrix(yt, pred))
```
